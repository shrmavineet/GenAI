```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader('../script.txt', encoding='utf-8')

texts = loader.load()
onDoc = texts[0]

print(onDoc.page_content)
```

#### PDF Document Loader

```python
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader('../JavaScript-Handwritten-Notes.pdf')

docs = loader.load()
len(docs)
onPage = docs[0]
print(onPage.page_content)
```
