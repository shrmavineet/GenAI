#### Prompt Templates - Dynamic Prompts

```python
from langchain_mistralai import ChatMistralAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatMistralAI(model = "mistral-small-latest")
out = StrOutputParser()

prompts = ChatPromptTemplate.from_messages([
    {"role":"system", "content":"You are translator and translate input in {language}"},
    {"role":"system", "content":"{query}"},
])

final_prompts = prompts.format_messages(language = "german", query = "I love Python, and JavaScript")

res = llm.invoke(final_prompts)
print(res)
```

```python
#### My Self Chain Node
def transform_case(result:str):
    return result.upper()
```

#### Chains - Runnable

```python
res =llm.invoke(prompts.invoke({"language":"Hindi", "query":"I love python language"}))
res.content
```

```python
chains = prompts | llm | out | transform_case

res = chains.invoke({"language":"Hinglish", "query":"I love LangChain"})

print(res)
```
