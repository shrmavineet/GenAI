```python
from pydantic import BaseModel, Field
from typing import List,Annotated
```

```python
class User(BaseModel):
    name: str
    age: int
```

```python
user = User(name="Vineet", age="22")
```

```python
print(user)
```

#### Field Vailidation

```python
class Products(BaseModel):
    pname: str = Field(max_length=10)
    des: str = Field(min_length=20)
    price: int = Field(gt=0, le=100)
```

```python
model = Products(pname="SamsungS23", des="This is amazing model form samsung...", price=99)
model2 = Products(pname="SamsungS24", des="This is amazing model form sams...", price=89)
```

```python
class Order(BaseModel):
    id:str = Field(min_length=6, max_length=20)
    product_details: Products
    
```

```python
order = Order(id='1234rr',product_details=model)
order
```

```python
class Carts(BaseModel):
    id:str = Field(min_length=6, max_length=12)
    items :List[Products]
    comment: str = Field(
        default="No Comment"
    )
```

```python
cart = Carts(id="asdfghq", items=[model, model2], comment="This is my cart")
cart
```

#### Annotated

```python
commonValidator = Annotated[List[Products], "This is Common Validator"]
```

```python
class Carts(BaseModel):
    id:str = Field(min_length=6, max_length=12)
    items :commonValidator
    comment: str = Field(
        default="No Comment"
    )
```
