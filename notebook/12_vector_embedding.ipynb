```python
from dotenv import load_dotenv
load_dotenv()
```

```python
documents = [
    "Machine learning is a subset of artificial intelligence",
    "Deep learning uses nureal networks with multiple layes",
    "Python is a programming language popular for data science",
    "India won the cricket world cup in 2011",
    "AI model can understand and generate human language"
]
```

```python
from langchain_mistralai import MistralAIEmbeddings

embedding = MistralAIEmbeddings(model="mistral-embed-2312")

vector = embedding.embed_documents(documents)

len(vector)
```

#### Vector DB

```python
from langchain_chroma import Chroma

vector_store = Chroma.from_texts(
    texts = documents,
    embedding=embedding,
    persist_directory="./vector_db"
)
```

```python
query = "Python"

result = vector_store.similarity_search(query=query,k=2)
```
