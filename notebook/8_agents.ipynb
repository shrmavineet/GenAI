```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool

@tool
def number_add(a: int, b:int):
    """
        This function return the sum of two numbers
        Args:
            a: First Number
            b: Second Number
    """
    return a+b

@tool
def number_multiplication(a:int, b:int):
    """
        This function return the multiplication of two number
        Args:
            a: First Number
            b: Second Number
    """
    return a*b
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(model="openai/gpt-oss-20b")
llm
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model = llm,
    tools=[number_add,number_multiplication],
    system_prompt="You a math teacher anf always use tool for calculation"
)
response = agent.invoke({"messages":[{"role":"user","content":"What is the sum of 2+3?"}]})

response['messages'][-1].content
# for msg in response['messages']:
#     print(msg)
```
