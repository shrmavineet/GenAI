```python
from dotenv import load_dotenv
load_dotenv()

from langchain_mistralai import ChatMistralAI
llm = ChatMistralAI(model="mistral-small-latest")
```

```python
from langchain_core.tools import tool

@tool
def sub_num(a:int, b:int):
    """
        It return the subtraction of two number
        Args:
            a: Fisrt Number which i subtracte from second number
            b: Second Number which i subtract to first number
    """
    return a-b

@tool
def multi_num(a:int, b:int):
    """
        It return the multiplication of two number
        Args:
        a: First Number
        b: Second Number
    """
    return a*b

```

```python
from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    tools=[sub_num,multi_num],
    system_prompt="You are math expert and always use tools in case of calculation"
)

response = agent.invoke({"messages":[{"role":"user","content":"What is the multiplication 4 and 9 ?"}]})

response["messages"][-1].content
```
