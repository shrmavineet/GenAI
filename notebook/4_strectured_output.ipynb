```python
from dotenv import load_dotenv
load_dotenv()

from langchain_groq import ChatGroq

llm = ChatGroq(model = "openai/gpt-oss-20b")
```

### Discuss the Problem FIrst

```python
text = "Hello my name is Vineet Sharma" \
"My Email Address is vineetshrma7@gmail.com" \
"My age is 24"

res = llm.invoke(f"Please give only name, email and age from this text: {text}")

print(res.content)
print(type(res.content))
```

#### Solution is : Structured Output

```python
from pydantic import BaseModel, Field #define the structure of a value
from typing import List # return result in list

class ResponseStructure(BaseModel):
    name:str = Field(description="Complete Name")
    email:str = Field(description="Full email Address")
    age:int = Field(description="Age")

strucutred_llm = llm.with_structured_output(ResponseStructure)

result = strucutred_llm.invoke(f"Please give only name, email and age from this text: {text}")
result.model_dump()
print(result)
result.name
```

```python
class Movie(BaseModel):
    s_no:int = Field(description="Sequence wise Number")
    name:str = Field(description="Movie Name")
    yrs:int = Field(description="Movie Release date")
    rating:float = Field(description="Imdb Rating")

class All_Movies(BaseModel):
    movies:List[Movie]

movies_llm = llm.with_structured_output(All_Movies)

res = movies_llm.invoke("Give me a 5 Sci-Fi movies")
res
```
