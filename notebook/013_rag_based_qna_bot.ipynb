```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_mistralai import MistralAIEmbeddings
from langchain_chroma import Chroma
```

```python
loader = PyPDFLoader("../data/data_science_syllabus.pdf")
docs = loader.load()

splitter = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap = 200)

splitted_data = splitter.split_documents(docs)
```

```python
embedding = MistralAIEmbeddings(model = "mistral-embed-2312")
```

```python
vetor = Chroma.from_documents(
    documents=splitted_data,
    embedding=embedding
)
```

```python
query = "machine Learning and Data Science Content"

data = vetor.similarity_search(query=query)
data
```

```python
context = ""

for doc in data:
    context = context + doc.page_content + "\n"

print(context)
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(model = "openai/gpt-oss-20b")
```

```python
res = llm.invoke(f"Can you provide me the answer based on provided content for my question, context: {context} and question: {query}")
print(res.content)
```

#### [Making chain] Chain - Content_generate | prompt | llm | strparser

```python
def get_content(query:str):
    data = vetor.similarity_search(query=query)

    context = ""
    for docs in data:
        context = context + docs.page_content

    return{
        "context": context,
        "question": query
    }
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template(
    """
    You are a helpful assistant and provide answered based on the context fro user question and if you don't know the answer, then you can say that i'I dont know'
    context: {context},
    question {question}
""")
```

```python
rag_chain = get_content | prompt | llm

res = rag_chain.invoke("What is duration for javascript program?")
```

```python
print(res.content)
```
