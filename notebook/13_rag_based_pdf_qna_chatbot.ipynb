```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import  RecursiveCharacterTextSplitter
from langchain_mistralai import MistralAIEmbeddings,ChatMistralAI
from langchain_chroma import Chroma
from langchain_core.prompts import PromptTemplate
```

```python
loader = PyPDFLoader("../data/data_science_syllabus.pdf")
docs = loader.load()
len(docs)
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap=100)
splitted_data = splitter.split_documents(docs)
```

```python
embeddings = MistralAIEmbeddings(model="mistral-embed-2312")
```

```python
vector = Chroma.from_documents(
    documents = splitted_data,
    embedding = embeddings
)
```

```python
query= "Machine Learnings and Data Science Content"
data = vector.similarity_search(query=query)
```

```python
data
```

```python
context =""

for doc in data:
    context += doc.page_content + "\n"

print(context)
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(model="openai/gpt-oss-20b")

final_data = llm.invoke(f"Can you provide me the answer based on provided context for my question, context: {context} and question :{query}")

print(final_data.content)


# llm = ChatMistralAI(model="mistral-small-latest")

# final_data = llm.invoke(f"Can you provide me the answer based on provided context for my question, context: {context} and question :{query}")

# print(final_data)
```
